请参考作业 01，独立编写完整的 PyTorch 训练流程，让下面的预测函数尽量拟合给定数据。

本题只给出三个线索。参数如何创建、如何计算梯度以及如何更新，都由你完成。可以查阅作业 01，但请根据本题的预测函数调整代码。

## 线索 1：预测函数

这次有三个待学习参数 $w_1,w_2,w_3$：

$$\hat y=w_1x+w_2x^2+w_3.$$

请根据这个公式，自己编写计算预测值的代码。输入整个 `x` 张量时，应得到每个样本对应的预测值。三个待学习参数也由你创建。请保留这个预测公式。

其中，$w_1$ 是一次项的系数，$w_2$ 是二次项的系数，$w_3$ 是常数项。三个参数在训练中都需要被调整。

## 线索 2：损失函数

每个样本的损失仍然是预测误差的绝对值：

$$L_i=|y_i-\hat y_i|.$$

本次有多个样本，请将所有样本的损失取平均，得到用于训练的一个数：

$$L=\frac{1}{N}\sum_{i=1}^{N}|y_i-\hat y_i|.$$

这里 $N$ 是样本数。请自己用 PyTorch 写出这个损失，不要换成平方误差。

## 线索 3：两个数据张量

`x[i]` 和 `y[i]` 是同一个样本的输入和目标。训练时保持这些数据不变，不要直接填写你猜测的参数作为答案。

```python
import torch

# 输入数据与对应的目标数据，均为浮点数
x = torch.tensor([-2.0, -1.5, -1.0, -0.5, 0.0, 0.5, 1.0, 1.5, 2.0])
y = torch.tensor([-3.0, -1.375, 0.0, 1.125, 2.0, 2.625, 3.0, 3.125, 3.0])
```

## 你的作答区

请在下面编写完整训练代码，可以自行增加单元格。训练结束后，展示学到的三个参数、每个样本的预测值，以及最终的平均绝对误差。最终展示的预测值和损失应使用训练结束后的参数计算。

再用文字说明：你对作业 01 的训练代码做了哪些修改？你如何判断这次训练有效？不要求参数或误差恰好等于某个数，也不要求每一轮损失都下降。

```python
def train_three_params(epochs=1000, base_lr=0.1, decay=True):
    # 初始化三个待学习参数 w1, w2, w3
    w1 = torch.tensor(0.0, requires_grad=True)
    w2 = torch.tensor(0.0, requires_grad=True)
    w3 = torch.tensor(0.0, requires_grad=True)
    
    # 将三个参数传入优化器
    optimizer = torch.optim.SGD([w1, w2, w3], lr=base_lr)
    history = []

    for epoch in range(1, epochs + 1):
        # 动态调整学习率（和图1一样使用 0.1 / 轮数）
        lr = base_lr / epoch if decay else base_lr
        optimizer.param_groups[0]["lr"] = lr
        
        optimizer.zero_grad() # 清除旧梯度
        
        # --- 线索 1：预测函数 ---
        # \hat{y} = w1*x + w2*x^2 + w3
        y_pred = w1 * x + w2 * (x ** 2) + w3
        
        # --- 线索 2：损失函数 (平均绝对误差 MAE) ---
        # L = 1/N * Σ|y_i - y_hat_i|
        loss = torch.abs(y - y_pred).mean()
        
        loss.backward()      # 计算梯度
        optimizer.step()     # 更新参数
        
        # 记录更新后的结果，不构建求导过程
        with torch.no_grad():
            current_loss = torch.abs(y - y_pred).mean().item()
            # 保留部分轮次的历史记录用于打印
            if epoch in [1, 2, 5, 10, 100, 500, 1000]:
                history.append((epoch, lr, w1.item(), w2.item(), w3.item(), current_loss))
                
    return w1.detach().item(), w2.detach().item(), w3.detach().item(), history

# ----------------- 3. 运行训练并打印过程 (仿照图1的输出格式) -----------------
w1_final, w2_final, w3_final, history = train_three_params()

print("轮数\t学习率\t\tw1\t\tw2\t\tw3\t\t损失")
for epoch, lr, val1, val2, val3, current_loss in history:
    print(f"{epoch:4d}\t{lr:.6f}\t{val1:9.6f}\t{val2:9.6f}\t{val3:9.6f}\t{current_loss:.6f}")

print("-" * 80)

# ----------------- 4. 作答区要求：展示最终结果 -----------------
with torch.no_grad():
    # 使用训练结束后的参数计算最终预测值
    final_prediction = w1_final * x + w2_final * (x ** 2) + w3_final
    # 计算最终的平均绝对误差
    final_loss = torch.abs(y - final_prediction).mean().item()

print("【最终训练结果展示】")
print(f"学到的三个参数: w1 = {w1_final:.6f}, w2 = {w2_final:.6f}, w3 = {w3_final:.6f}")
print(f"\n每个样本的预测值:\n{final_prediction.numpy()}")
print(f"\n最终的平均绝对误差 (MAE): {final_loss:.6f}")
```
